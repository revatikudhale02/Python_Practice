```python
#slicing/accessing
my_list = ["apple","banana","mango"]
print(my_list[0])

# using reverse index
my_list[-2]
```

```python
#add
my_list = ["apple","banana","mango"]
my_list.append("strawberry")
my_list
```

```python
#insert
my_list = ['apple', 'banana', 'mango', 'strawberry']
my_list.insert(2,'Kiwi')
my_list
```

```python
#Add Multiple
my_List = [1,2,3]
List1 = [4,5,6]
my_List.extend(List1)
my_List
```

```python
#remmove specified item
my_list = ['apple', 'banana', 'mango', 'strawberry']
my_list.remove('banana')
my_list
```

```python
#remmove by index
my_list = ['apple', 'banana', 'mango', 'strawberry']
my_list.pop(2)
my_list
```

```python
#last item remove
my_list = ['apple', 'banana', 'mango', 'strawberry']
my_list.pop()
my_list
```

```python
#clear list
my_List.clear()
my_List
```

```python
#copying one list to other
my_list = ['apple', 'banana', 'mango', 'strawberry']
newList = my_list.copy()
newList
```

```python
#sort and copy
myList = [30,10,20]
newList = sorted(myList)
newList
```
