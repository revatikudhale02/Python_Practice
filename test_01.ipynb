```python
# Find the matching element from two lists
List_01 = ["Red","Yellow","Lightgreen"]
List_02 = [ "White","Black","Yellow"]

for i in List_01:
    for j in List_02:
        if i==j:
            print("Matched",i,j)
```

```python
List_01 = ["Red","Yellow","Lightgreen"]
List_02 = [ "White","Black","Yellow"]
List_03 = ["White","Purple","Yellow"]

for i in List_01:
    for j in List_02:
        for k in List_03:
            if i==j==k:
                print("Matched:",i,j,k)
```
