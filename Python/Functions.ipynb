```python
#With argument
def printSquares(list_01):
    NewDict_02 =""
    NewDict_02 = {}
    for i in list_01:
        NewDict_02[i]=i*i
    print(NewDict_02)
```

```python
#using the function
list_02 = [1,2,3,4]
printSquares(list_02)
```

```python
# list_01 = [1,2,3]
#dict comprehension
def printSquares(list_01):
    NewDict_02 = {i:i*i for i in list_01}
    print(NewDict_02)
```

```python
list_02 = [1,2,3]
printSquares(list_02)
```

```python
# update the dict

dict_01 ={"name":"Revati","Age":20}
dict_02 ={"Pin":411003,"city":"Pune"}

def addDict(dict1,dict2):
    newDict={**dict1,**dict2}
    print(newDict)

addDict(dict_01,dict_02)    
```

```python
#update the dictionary using function

Dict_01={"name":"Rajesh","age":24}
Dict_02={"Pin":444001,"Area":"Sanghavi"}

def updatelists(Dict_01,Dict_02):
    Dict_01.update(Dict_02)
    print(Dict_01)
updatelists(Dict_01,Dict_02)
```

```python
#create new list with values less than 5

list01 = [12,2,1,11,5,3]

def sort_list(list01):
    sortedlist = []
    for i in list01:
        if i < 5:
            sortedlist.append(i)
    return sortedlist

sort_list(list01)
```
