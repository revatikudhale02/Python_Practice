```python
#Access item in tuple
myTuple = ("apple", "banana", "mango")
myTuple[1]
```

```python
#slicing
myTuple = ("apple", "banana", "mango")
print(myTuple[::-1])

myTuple[1:2]
```

```python
#index 
myTuple = ("apple", "banana", "mango")
myTuple.index("banana")
```

```python
#count
myTuple = ("apple", "banana", "mango", "grapes")
myTuple.count("mango")
```

```python
#add/join
tuple1 = ("apple", "banana")
tuple2 = ("mango", "orange")
newTuple = tuple1 + tuple2
print(newTuple)

myTuple = ("apple", "banana")
myTuple += ("mango",)
myTuple
```
